```python
# =============================================================
# PRUEBA DE DESCARTE - Señal El Niño y heladas en Oruro
# Correr en Google Colab: pegar todo en UNA celda y presionar ▶
# =============================================================
import requests
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# ---------- 1. CONFIGURACIÓN (cambiar cuando sepamos el municipio) ----------
LAT = -17.83   # Obrajes, redondeado a 0,01° por privacidad
LON = -67.00
NOMBRE = "Obrajes"
UMBRAL_HELADA = -2.5   # °C: papa común sin aclimatar (Chen y Li 1980)
```

```python
# ---------- 2. BAJAR DATOS DE NASA POWER (diarios desde 1981) ----------
url = ("https://power.larc.nasa.gov/api/temporal/daily/point"
       "?parameters=T2M_MIN,PRECTOTCORR&community=AG"
       f"&longitude={LON}&latitude={LAT}"
       "&start=19810101&end=20261110&format=JSON")
print("Descargando NASA POWER... (puede tardar 30-60 s)")
datos = requests.get(url, timeout=180).json()["properties"]["parameter"]

df = pd.DataFrame({
    "tmin": datos["T2M_MIN"],
    "lluvia": datos["PRECTOTCORR"],
})
df.index = pd.to_datetime(df.index, format="%Y%m%d")
df = df.replace(-999, float("nan"))          # -999 = dato faltante
df.to_csv(f"power_{NOMBRE}.csv")             # guardamos copia (Plan B offline)
print("Días descargados:", len(df), "| último día:", df.index.max().date())
```

```python
# ---------- 3. BAJAR ÍNDICE ONI (NOAA) para saber qué años fueron Niño ----------
oni = pd.read_csv("https://www.cpc.ncep.noaa.gov/data/indices/oni.ascii.txt",
                  sep=r"\s+")
oni = oni[oni["SEAS"] == "NDJ"]               # trimestre nov-dic-ene
oni = oni.set_index("YR")["ANOM"]

def fase(valor):
    if valor >= 0.5:
        return "Niño"
    if valor <= -0.5:
        return "Niña"
    return "Normal"
```

```python
# ---------- 4. LLUVIA POR CAMPAÑA (noviembre a marzo) ----------
filas = []
for anio in range(1981, 2026):
    inicio = f"{anio}-11-01"
    fin = f"{anio + 1}-03-31"
    tramo = df.loc[inicio:fin]
    if len(tramo) < 140 or anio not in oni.index:
        continue
    filas.append({
        "campaña": f"{anio}-{str(anio + 1)[2:]}",
        "fase": fase(oni[anio]),
        "lluvia_mm": round(tramo["lluvia"].sum()),
        "dias_helada": int((tramo["tmin"] <= UMBRAL_HELADA).sum()),
    })
camp = pd.DataFrame(filas)
camp.to_csv(f"campañas_{NOMBRE}.csv", index=False)

print("\n=== PROMEDIO POR FASE (campaña nov-mar) ===")
print(camp.groupby("fase")[["lluvia_mm", "dias_helada"]].mean().round(1))
print("\nCantidad de campañas por fase:")
print(camp["fase"].value_counts())
```

```python
# ---------- 5. HELADAS EN FEBRERO ----------
feb = df[df.index.month == 2]
anios_con_helada = feb[feb["tmin"] <= UMBRAL_HELADA].index.year.unique()
total_anios = feb.index.year.nunique()
print(f"\nFebreros con al menos una helada <= {UMBRAL_HELADA} °C:",
      f"{len(anios_con_helada)} de {total_anios}",
      f"({100 * len(anios_con_helada) / total_anios:.0f} %)")
```

```python
# ---------- 6. ¿APARECE LA HELADA DEL 3 DE FEBRERO DE 2026? ----------
print("\nTemperatura mínima 1-5 de febrero de 2026:")
print(df.loc["2026-02-01":"2026-02-05", "tmin"])
```

```python
# ---------- 7. GRÁFICO ----------
colores = {"Niño": "tab:red", "Niña": "tab:blue", "Normal": "tab:gray"}
plt.figure(figsize=(12, 4))
plt.bar(camp["campaña"], camp["lluvia_mm"],
        color=[colores[f] for f in camp["fase"]])
plt.xticks(rotation=90)
plt.ylabel("Lluvia nov-mar (mm)")
plt.title(f"{NOMBRE}: lluvia por campaña (rojo = Niño, azul = Niña) - NASA POWER")
plt.tight_layout()
plt.savefig(f"lluvia_campañas_{NOMBRE}.png", dpi=150)
plt.show()
```

```python
# ---------- 8. NIÑO FUERTE (ONI >= 1.5) Y MÉTODO ----------
camp["oni"] = [oni[int(c[:4])] for c in camp["campaña"]]
promedio_normal = camp[camp["fase"] == "Normal"]["lluvia_mm"].mean()
camp["vs_normal_%"] = ((camp["lluvia_mm"] / promedio_normal - 1) * 100).round()
fuertes = camp[camp["oni"] >= 1.5]
print("Normal de referencia (promedio campañas Normales):", round(promedio_normal), "mm, n =", (camp["fase"] == "Normal").sum())
print(fuertes[["campaña", "oni", "lluvia_mm", "vs_normal_%"]])
secas = (fuertes["lluvia_mm"] < promedio_normal).sum()
print(f"\nNiño fuerte con lluvia bajo lo normal: {secas} de {len(fuertes)}")
camp.to_csv(f"campañas_{NOMBRE}.csv", index=False)
```

## 9. P6-4 · Contraste con CHIRPS (segunda fuente)

```python
# P6-4 · Contraste CHIRPS vs NASA POWER (corre DESPUÉS de las celdas 1-8: usa `camp` y `oni`)
# Fuente 1: NOAA ERDDAP (CHIRPS desde 1984). Fuente 2 si falla: IRI Data Library (Columbia).
import pandas as pd

LAT_PUB, LON_PUB = -17.83, -67.00   # punto publicado (redondeado por privacidad)

def bajar_erddap():
    url = ("https://coastwatch.pfeg.noaa.gov/erddap/griddap/chirps20GlobalDailyP05.csv?"
           f"precip%5B(1984-01-01):1:(2026-03-31)%5D%5B({LAT_PUB})%5D%5B({LON_PUB})%5D")
    d = pd.read_csv(url, skiprows=[1])
    d["time"] = pd.to_datetime(d["time"]).dt.tz_localize(None).dt.normalize()
    return d.set_index("time")["precip"], "NOAA ERDDAP"

def bajar_iri():
    url = ("https://iridl.ldeo.columbia.edu/SOURCES/.UCSB/.CHIRPS/.v2p0/.daily-improved/.global/.0p05/.prcp/"
           f"X/{LON_PUB}/VALUE/Y/{LAT_PUB}/VALUE/T/(1%20Jan%201981)/(31%20Mar%202026)/RANGEEDGES/"
           "T/exch/table-/text/text/skipanyNaN/-table/.csv")
    d = pd.read_csv(url)
    d.columns = ["time", "precip"] + list(d.columns[2:])
    d["time"] = pd.to_datetime(d["time"], format="mixed", dayfirst=True)
    return d.set_index("time")["precip"].astype(float), "IRI Data Library"

ch = None
for f in (bajar_erddap, bajar_iri):
    try:
        print("Probando", f.__name__, "... (1-2 min)")
        ch, fuente = f()
        break
    except Exception as e:
        print("  falló:", type(e).__name__, str(e)[:120])
if ch is None:
    print("\nESTADO P6-4: CHIRPS unavailable (ninguna fuente respondió). No bloquea el núcleo.")
else:
    ch.to_csv("chirps_Obrajes.csv")
    print(f"\nFuente: {fuente} | días descargados: {len(ch)} | desde {ch.index.min().date()} hasta {ch.index.max().date()}")
    filas = []
    for anio in range(1981, 2026):
        tramo = ch.loc[f"{anio}-11-01":f"{anio+1}-03-31"]
        if tramo.notna().sum() < 140:          # misma regla que POWER
            continue
        filas.append({"campaña": f"{anio}-{str(anio+1)[2:]}", "chirps_mm": round(tramo.sum(), 1)})
    chc = pd.DataFrame(filas)
    comp = camp.merge(chc, on="campaña")
    print("\n=== POWER vs CHIRPS, campañas nov-mar (periodo común) ===")
    print("Campañas comparadas:", len(comp), "| excluidas por < 140 días:", 45 - len(chc))
    print("Correlación de totales:", round(comp["lluvia_mm"].corr(comp["chirps_mm"]), 2))
    print("Diferencia media CHIRPS - POWER:", round((comp["chirps_mm"] - comp["lluvia_mm"]).mean(), 1), "mm")
    normal_ch = comp[comp["fase"] == "Normal"]["chirps_mm"].mean()
    normal_pw = comp[comp["fase"] == "Normal"]["lluvia_mm"].mean()
    fuertes = comp[comp["oni"] >= 1.5]
    print("\nNormal CHIRPS:", round(normal_ch), "mm | Normal POWER (mismo periodo):", round(normal_pw), "mm")
    print(fuertes[["campaña", "oni", "lluvia_mm", "chirps_mm"]])
    print("Niño fuerte bajo la normal según CHIRPS:", (fuertes["chirps_mm"] < normal_ch).sum(), "de", len(fuertes))
    print("Niño fuerte bajo la normal según POWER (mismo periodo):", (fuertes["lluvia_mm"] < normal_pw).sum(), "de", len(fuertes))
    comp.to_csv("contraste_POWER_CHIRPS_Obrajes.csv", index=False)
```
